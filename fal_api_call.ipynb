```python


def get_model_gist_from_12labs(video_id):
  '''
  Needs 12labs api key stored in environment variables

  1. Steps create index on 12 labs
  2. Upload video to that index
  3. Let it process
  4. Fetch gist from the uploaded video

  '''
  from twelvelabs import TwelveLabs
  from twelvelabs.indexes import IndexesCreateRequestModelsItem
  from twelvelabs.tasks import TasksRetrieveResponse

  labs12_api = userdata.get('labs12')
  client = TwelveLabs(api_key=labs12_api)
  print("Connected to TwelveLabs")

  # 5. Generate title, topics, and hashtags
  gist = client.gist(video_id=video_id,types=["title", "topic", "hashtag"])

  # 6. Process the results
  print(f"Title={gist.title}\nTopics={gist.topics}\nHashtags={gist.hashtags}")
```
